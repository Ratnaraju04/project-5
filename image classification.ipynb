```python
import streamlit as st
import numpy as np
import tensorflow as tf
from PIL import Image

# ----------------------------
# Load Model
# ----------------------------
@st.cache_resource
def load_model():
    model = tf.keras.models.load_model("XRay_Model.keras")
    return model

model = load_model()

# Class Labels (Same as Training)
class_labels = {
    0: "NORMAL",
    1: "PNEUMONIA",
    2: "TUBERCULOSIS",
    3: "UNKNOWN"
}

# ----------------------------
# Streamlit UI
# ----------------------------
st.title("🩻 X-Ray Disease Prediction System")
st.write("Upload Chest X-Ray Image (64x64 grayscale trained model)")

uploaded_file = st.file_uploader("Choose an X-ray image...", type=["jpg", "jpeg", "png"])

if uploaded_file is not None:
    
    # Show Image
    image = Image.open(uploaded_file)
    st.image(image, caption="Uploaded Image", use_column_width=True)
    
    # Preprocess Image (Same as your notebook)
    image = image.convert("L")
    image = image.resize((64, 64))
    image = np.array(image)
    image = image.astype("float32") / 255.0
    image = np.expand_dims(image, axis=0)

    # Prediction
    prediction = model.predict(image)
    predicted_class = np.argmax(prediction)
    confidence = np.max(prediction) * 100

    st.subheader("Prediction Result")
    st.success(f"Predicted Disease: {class_labels[predicted_class]}")
    st.info(f"Confidence: {confidence:.2f}%")
```
